The Python for this workshop is very minimal and is given below. It consists of a single simple block of code that uses MQTT to read messages posted to two topics - currently joetopic1 and joetopic2 - on the Mosquitto test server.

```python
# Based off this: https://github.com/eclipse/paho.mqtt.python/blob/master/examples/subscribe_simple.py
import paho.mqtt.subscribe as subscribe

topics = ['joetopic1', 'joetopic2']

data = []
m = subscribe.simple(topics, hostname="pf-c0uudk9ksjw7kkm9chm5.cedalo.cloud", retained=False, msg_count=len(topics))

# If you just have a single topic you're subscribing to, use this:
# print(m.payload)

# If you're subscribing to multiple topics, use this:
for a in m:
        data.append(float(a.payload))

print(data)
```
